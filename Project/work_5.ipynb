Bài tập 5.2: Xây dựng một mạng CNN để nhận dạng chữ số bằng dataset MNIST (Số lớp và các tham số channel, kernel size, stride, padding tùy chọn). 
Yêu cầu: đầu vào là ảnh kích thước 2x2. Đầu ra có 10 neural ứng với 10 chữ số. Mỗi lớp sẽ bao gồm nn.Conv2d() và nn.ReLU().
Lưu ý, sửa in_features của lớp Linear sao cho bằng tổng số neural của lớp trước đó.


```python
from utils import *
torch.manual_seed(1)
np.random.seed(1)

class MyCNN(nn.Module):
    def __init__(self):
        super().__init__()

        # Insert code here

        # End code here
        self.flat = nn.Flatten()
        self.linear = nn.Linear(in_features=28*28,out_features=10)
        self.relulinear = nn.ReLU()
        
    def forward(self,x):

        # Insert code here

        # End code here

        x = self.relulinear(self.linear(self.flat(x)))
        return x

```

```python
train_dataset = datasets.MNIST(root = "dataset/", train=True, transform=transforms.ToTensor(), download=True)
test_dataset = datasets.MNIST(root = "dataset/", train=False, transform=transforms.ToTensor(), download=True)

batch_size = 32
num_epochs = 10
lr = 1e-3
train_loader = DataLoader(dataset=train_dataset, batch_size=batch_size, shuffle=True)
test_loader = DataLoader(dataset=test_dataset, batch_size=batch_size, shuffle=True)

model_CNN = MyCNN()
CELoss = nn.CrossEntropyLoss()
optimizer_CNN = optim.SGD(model_CNN.parameters(), lr = lr)
```

```python
trained_model, CNN_acc_list = trainer_CNN(model_CNN, train_loader, test_loader, num_epochs, batch_size, optimizer_CNN, CELoss)
print(f'Linear NN with 2 layers: Train_acc: {CNN_acc_list[-1][0]:.2f}%. Test_acc: {CNN_acc_list[-1][1]:.2f}%.')
plot_curve(CNN_acc_list, curve_type='both')
```
